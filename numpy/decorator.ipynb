```python
#decorator are used to pass an func as a argemt in another func
```

```python
def new_decorator(func):
    def wrapper_func():
        print('this is before decorater give')
        func()
        print("This is after decorater given")
    return wrapper_func

def dec_needed():
    print("This is a decorater function ")
```

```python
dec_needed()
```

```python
new = new_decorator(dec_needed)
```

```python
new()
```

```python
@new_decorator
def dec_needed():
    print("I'm not decoarated")
```

```python
def trial_decorator(func):
    def wrapper_func():
        print('this is before decorater give')
        func()
        print("This is after decorater given")
    return wrapper_func
```

```python
@trial_decorator
def trial_needed():
    print("I'm not decorator")
    
```

```python
trial_needed()
```
