```python
y = df['Result'].map({'Fail': 0,'Pass': 1})
#y = df['Result']
```

```python
x = pd.get_dummies(x, drop_first=True)
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y,test_size=0.2,random_state=8,stratify=y)
```

```python
print(xtest)
print (ytest)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
xtrain = scaler.fit_transform(xtrain)
xtest = scaler.transform(xtest)
```

```python
print(xtrain)
print (xtest)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=22000,n_jobs=-1)
```

```python
model1=model.fit(xtrain, ytrain)
ypred = model1.predict(xtest)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,classification_report
```

```python
accuracy_score(ytest,ypred)
```

```python
precision_score(ytest,ypred)
```

```python
recall_score(ytest,ypred)
```

```python
pd.crosstab(
    ytest,
    ypred,
    rownames=["Actual"],
    colnames=["Predicted"]
)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model_rf = RandomForestClassifier(
    n_estimators=100,
    random_state=8
)

model_rf.fit(xtrain, ytrain)

y_pred_rf = model_rf.predict(xtest)
```

```python
from sklearn.metrics import accuracy_score, classification_report

print("Accuracy:", accuracy_score(ytest, y_pred_rf))
print(classification_report(ytest, y_pred_rf))
```
