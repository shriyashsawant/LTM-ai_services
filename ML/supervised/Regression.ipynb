```python
y = df["medv"] #dependent vzriable and what to predict #here we assign the y for training
x=df.iloc[:,:13] #all the data other than medv #and x for what it will train on to predict
```

```python
#Distribution
sns.displot(df.medv)
```

```python
from sklearn.model_selection import train_test_split
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.25,random_state=8) #divide variable in to training set , i've choosed 25% training set 
#here random state is a tag give to the random split done to make it consistent across
```

```python
sns.distplot(ytrain) #here we can see that the training and test data is splitted in equal plot thus the training is correct
```

```python
sns.distplot(ytest) #both train and test data is same 
```

```python
xtrain[:5]
```

```python
from sklearn.linear_model import LinearRegression 
```

```python
lr = LinearRegression() #imported liner regression
```

```python
model1=lr.fit(xtrain,ytrain) #this trained the model on liner regression
```

```python
model1 #the model is created now
```

```python
model1.intercept_ #this is when all values passed would be 0 which would be b0
```

```python
model1.coef_  #this will give the slope of each attrivute according to the data is was trained on
```

```python
model1.predict(xtest[:1]) #this is the linear predicted value
```

```python
pred = model1.predict(xtest) 
```

```python
from sklearn.metrics import mean_squared_error
```

```python
err = mean_squared_error(ytest,pred)
```

```python
err = pd.DataFrame({"Actual":ytest,"Predict":pred})
```

```python
err
```

```python
err.reset_index(drop=True,inplace=True)
err
```

```python
plt.plot(err)
plt.legend(["Actual","Predicted"])
```

```python
from sklearn.metrics import r2_score
r2_score(ytest, pred)
```
