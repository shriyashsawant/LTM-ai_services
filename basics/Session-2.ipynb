```python
#break,pass,continue
#break - break out of the loop
#pass - do nothing
#continue - skip the current value

n=0
for n in range(10):
    if n==5:
        print(n)
        pass


for n in range(10):
    if n==5:
        print(n)
        break

for n in range(10):
    if n==5:
        print(n)
        continue
```

```python
n = int(input("Enter number of students: "))

for i in range(n):
    print(f"\nStudent {i+1}")

    sub1 = int(input("Enter marks of subject 1: "))
    sub2 = int(input("Enter marks of subject 2: "))
    sub3 = int(input("Enter marks of subject 3: "))

    avg = (sub1 + sub2 + sub3) / 3

    print("Average:", avg)

    if avg > 75:
        print("Qualified")
    else:
        print("Disqualified")
#ask for marks of 3 subs, then cal average of those 3 subject for each student, print qualified if avg >75, else disqualified
```

```python
students = {}

n = int(input("Enter number of students: "))

for i in range(n):
    name = input("Enter student name: ")

    marks = []

    for j in range(3):
        mark = int(input(f"Enter marks for subject {j+1}: "))
        marks.append(mark)

    students[name] = marks

print("\nStudent Results")

for name, marks in students.items():
    avg = sum(marks) / len(marks)

    print(f"\nName: {name}")
    print(f"Marks: {marks}")
    print(f"Average: {avg}")

    if avg > 75:
        print("Qualified")
    else:
        print("Disqualified")
```
