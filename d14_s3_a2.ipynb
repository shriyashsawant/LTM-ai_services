```python

import numpy as np


# 1. Decision Values
def decision_values(X, w, b):

    X = np.asarray(X)
    w = np.asarray(w)

    return np.dot(X, w) + b


# 2. Predict
def predict(X, w, b):

    values = decision_values(X, w, b)

    return np.where(values >= 0, 1, -1)


# 3. Geometric Margin
def geometric_margin(w):

    w = np.asarray(w)

    norm_w = np.linalg.norm(w)

    if norm_w == 0:
        raise ValueError("Weight vector cannot be zero")

    return float(2.0 / norm_w)


# 4. Distance to Hyperplane
def distance_to_hyperplane(X, w, b):

    X = np.asarray(X)
    w = np.asarray(w)

    norm_w = np.linalg.norm(w)

    if norm_w == 0:
        raise ValueError("Weight vector cannot be zero")

    values = decision_values(X, w, b)

    return np.abs(values) / norm_w


# Verification
rng = np.random.default_rng(0)

X = rng.normal(size=(6, 3))

w = np.array([1.0, -2.0, 0.5])

b = 0.75


# Test 1: Decision Values
print("Decision Values:")
print(decision_values(X, w, b))

print("Shape:", decision_values(X, w, b).shape)


# Test 2: Predictions
print("\nPredictions:")
print(predict(X, w, b))


# Test 3: Geometric Margin
print("\nGeometric Margin:")
print(geometric_margin(w))

print("Expected Margin:")
print(round(geometric_margin(np.array([3.0, 4.0])), 3))


# Test 4: Distance to Hyperplane
print("\nDistance to Hyperplane:")
print(distance_to_hyperplane(X, w, b))

print("\nDistance of Origin:")
print(distance_to_hyperplane(
    np.array([[0, 0, 0]]), w, b
))


# Additional Verification
print("\nScale Invariance:")

w_scaled = 2 * w
b_scaled = 2 * b

print("Original Distance:")
print(distance_to_hyperplane(X, w, b))

print("Scaled Distance:")
print(distance_to_hyperplane(X, w_scaled, b_scaled))


print("\nMargin Comparison:")
print("Original:", geometric_margin(w))
print("Doubled Weight:", geometric_margin(2 * w))
```
