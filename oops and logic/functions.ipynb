```python
def is_prime(n):
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

is_prime(6)            


```

```python
for i in range(2,11):

    if (is_prime(i)==True):
        print(i)
```

```python
n = int(input("Provide the number: "))
prime_count = []
count = 0

while count < 5:
    if is_prime(n):
        prime_count.append(n)
        count += 1
    n += 1

for num in prime_count:
    print(num)

```

```python
class Employee:
    '''This class provides rules to create employee records'''

    org = 'LTM' #global variable for class
    Loc = 'Kalinga'
    City = 'BBSR'
    count = 0

    def __init__(self,id,name,salary): #this is constructor in python 
        self.name = name
        self.id = id
        self.salary = salary
        Employee.count += 1

    def __repr__(self):#here repr is a already pre defined class in pythin we're overiding it to our needs
        return __class__.__name__,Employee.count #the funtion return value which is define inside the class 
    #thus when a object is intialized form this calls it'll follow this repr

    e3 = Employee(3,"Sean",40000)
    print(e3.__repr__()) #it'll return the class name which is employee and count and employee count
```
