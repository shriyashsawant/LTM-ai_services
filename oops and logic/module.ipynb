```python
class CustomGen:
    # Constructor runs when an object is created
    def __init__(self, limit):
        self.current = 1      # Starting value
        self.limit = limit    # Maximum value to generate

    # Generator method
    def next(self):
        # Keep generating values until current exceeds limit
        while self.current <= self.limit:
            yield self.current      # Return current value and pause
            self.current += 1       # Move to next value


# Create object
g = CustomGen(5)

# Call generator method and iterate over values
for num in g.next():
    print(num)
                 
```
