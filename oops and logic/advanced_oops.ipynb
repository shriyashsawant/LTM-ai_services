```python
class car:
    def move(self):
        return('drive on the road')

class boat:
    def move(self):
        return('cruise in the water')

#Same function but different class
```

```python
def start_journey(vehicle):
    print(vehicle.move())  #this print depending on the vehicle passed to it either car or boat

    #thus polymorphism is achieved 
```

```python
start_journey(car())
```

```python
from abc import ABC,abstractmethod #abc = abstract base class ,abstract method is the decorator
```

```python
class football(ABC):
    @abstractmethod # this is the abstract decorator to define the implementation is abstract
    def kick(self):  #this is the abstract class made for later development
        pass

    @abstractmethod
    def passed(self):   # thus when any class inheritcs this football class it is compullsory to define these two classes
        pass

class fifa27(football):
    def kick(self):
        print("messi kicked the ball")
    def passed(self):
            print("messi scored no need to pass")

f=fifa27()
f.passed()
```

```python
class book:
    def __init__(self,title,author,pages):
        self.title = title
        self.author = author
        self.pages = pages

    def __str__(self):
        print("The {} is written by {} and has {} pages".format(self.title,self.author,self.pages))

    def __len__(self):
        return self.pages

    def __del__(self):
        print("The book is deleted")
```

```python
b1 = book('King','Shriyash',500)
```

```python
b1.__str__()
```

```python
b1.__len__()
```

```python
b1.__del__()
```

```python
class student(person):
    def check_eligible(self):
        if self.age>21:
            return True
        else :
            return False

    def enroll(self,roll_no):

        if roll_no != self.roll_no:
            return "Roll no not found"
    
        self.roll_no = roll_no
        if self.check_eligible() ==True:
            selected_course = input(
                f"Choose courses from {t1.get_courses()}: ").split(",")
            self.student_course = {self.roll_no: selected_course}
            

            return f'roll no {self.roll_no} is sucessfully enrolled to {self.student_course} courses'

        
s1= student('shriyash','sawant',22,300)
s1.enroll(300)
```
