```python

import numpy as np

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score


# 1. Majority Vote
def majority_vote(preds):

    preds = np.array(preds)

    n_samples = preds.shape[1]

    final_predictions = []

    for i in range(n_samples):

        votes = np.bincount(preds[:, i])

        final_predictions.append(np.argmax(votes))

    return np.array(final_predictions)


# 2. Forest Accuracy
def forest_accuracy(
    X_train, y_train, X_test, y_test,
    n_estimators, max_features, seed=0
):

    model = RandomForestClassifier(
        n_estimators=n_estimators,
        max_features=max_features,
        random_state=seed
    )

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    return round(float(accuracy_score(y_test, y_pred)), 4)


# 3. Estimator Sweep
def estimator_sweep(
    X_train, y_train, X_test, y_test,
    n_list, seed=0
):

    results = {}

    for n in n_list:

        accuracy = forest_accuracy(
            X_train,
            y_train,
            X_test,
            y_test,
            n_estimators=n,
            max_features="sqrt",
            seed=seed
        )

        results[n] = accuracy

    return results


# 4. Feature Importances Top K
def feature_importances_topk(X_train, y_train, k, seed=0):

    model = RandomForestClassifier(
        n_estimators=200,
        random_state=seed
    )

    model.fit(X_train, y_train)

    importances = model.feature_importances_

    top_indices = np.argsort(importances)[::-1][:k]

    return top_indices.astype(int)


# Verification
d = load_wine()

X_train, X_test, y_train, y_test = train_test_split(
    d.data,
    d.target,
    test_size=0.25,
    stratify=d.target,
    random_state=5
)


# Test 1: Majority Vote
preds = np.array([
    [2, 0, 1, 0],
    [0, 1, 1, 2],
    [0, 1, 2, 2]
])

print("Majority Vote:")
print(majority_vote(preds))


# Test 2: Forest Accuracy
print("\nForest Accuracy:")
print(
    forest_accuracy(
        X_train, y_train,
        X_test, y_test,
        100, "sqrt", 0
    )
)


# Test 3: Estimator Sweep
print("\nEstimator Sweep:")
print(
    estimator_sweep(
        X_train, y_train,
        X_test, y_test,
        [1, 5, 100], 0
    )
)


# Test 4: Feature Importance
print("\nTop 5 Feature Indices:")
print(
    feature_importances_topk(
        X_train, y_train, 5, 0
    )
)
```
